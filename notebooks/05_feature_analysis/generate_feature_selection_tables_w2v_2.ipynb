```python
import pandas as pd
```

```python
df_100 = pd.read_csv("../../results/tables/source/fs_w2v_2_top100.tsv", sep="\t")
df_100.head()
```

```python
df_104 = pd.read_csv("../../results/tables/source/fs_w2v_2_top104.tsv", sep="\t")
df_104.head()
```

```python
df_100f = df_104.head(100)
df_100f.shape
```

```python
for col in df_100.columns:
    comparison = df_100[col] == df_100f[col]
    mismatches = (~comparison).sum()
    print(f"Column '{col}': {mismatches} mismatches")
```
